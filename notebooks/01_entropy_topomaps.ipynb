# Topomap spectral entropy

Visualizzazione della spectral entropy sui 27 canali. Le finestre vengono aggregate entro soggetto prima del confronto HC/MS, evitando pseudoreplicazione.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display
from ipywidgets import Dropdown, IntSlider, interact

from eeg_ms.config import CANONICAL_DATA, CHANNELS_FILE, FIGURES
from eeg_ms.visualization.topomaps import (
    load_eeg_info,
    plot_condition_contrasts,
    plot_group_comparison,
    plot_topomap,
    subject_channel_map,
)

TOPOMAP_FIGURES = FIGURES / "topomaps"
TOPOMAP_FIGURES.mkdir(parents=True, exist_ok=True)

canonical = pd.read_parquet(CANONICAL_DATA)
info = load_eeg_info(CHANNELS_FILE)

print(f"Righe canoniche: {len(canonical):,}")
print(f"Soggetti: {canonical['subject_id'].nunique()}")
print(f"Canali nel montage: {len(info.ch_names)}")
```

## Controllo dei valori disponibili

```python
entropy = canonical.loc[
    canonical["measure"].eq("spectral_entropy")
    & canonical["spatial_level"].eq("channel")
]

display(entropy.groupby(["group", "condition"])["value"].agg(["count", "mean", "std", "min", "max"]))
```

## Entropy HC/MS in CE e OE

```python
for condition in ["CE", "OE"]:
    figure = plot_group_comparison(
        canonical,
        info,
        condition=condition,
        measure="spectral_entropy",
        band="broadband",
    )
    figure.savefig(TOPOMAP_FIGURES/f"spectral_entropy_{condition}_hc_ms.png", dpi=200, bbox_inches="tight",)
    plt.show()
```

## Differenza appaiata CE-OE

```python
figure = plot_condition_contrasts(
    canonical,
    info,
    measure="spectral_entropy",
    band="broadband",
)
figure.savefig(
    TOPOMAP_FIGURES/"spectral_entropy_paired_CE_minus_OE.png",
    dpi=200,
    bbox_inches="tight",
)
plt.show()
```

## Esplorazione interattiva per soggetto e finestra

`window=0` indica la mediana sulle finestre.

```python
subject_options = sorted(canonical["subject_id"].unique())

@interact(
    subject_id=Dropdown(options=subject_options, description="Soggetto"),
    condition=Dropdown(options=["CE", "OE"], description="Condizione"),
    window=IntSlider(min=0, max=18, value=0, description="Finestra"),
)
def explore_entropy(subject_id, condition, window):
    values = subject_channel_map(
        canonical,
        info,
        subject_id=subject_id,
        condition=condition,
        measure="spectral_entropy",
        band="broadband",
        window=None if window == 0 else window,
    )
    title_window = "mediana finestre" if window == 0 else f"finestra {window}"
    figure, _ = plot_topomap(
        values,
        info,
        title=f"{subject_id} · {condition} · entropy · {title_window}",
        cmap="magma",
        vlim=(0.0, 1.0),
    )
    plt.show()
```

## Interpretazione

Una spectral entropy maggiore indica una distribuzione spettrale più uniforme; valori inferiori indicano maggiore concentrazione della potenza in specifiche frequenze. La scala 0–1 è giusta perché le feature presenti nel dataset risultano normalizzate in tale intervallo.

## Variabilità tra soggetti: spectral entropy

Per ogni soggetto si calcola prima la mediana delle finestre; la mappa mostra poi la deviazione standard campionaria tra soggetti, canale per canale. HC e MS condividono la scala cromatica. La differenza `MS - HC` descrive dove la dispersione è maggiore, ma non è un test di significatività.

```python
from eeg_ms.visualization.topomaps import group_channel_std_map, plot_topomap

for condition in ["CE", "OE"]:
    hc_std = group_channel_std_map(
        canonical,
        info,
        group="hc",
        condition=condition,
        measure="spectral_entropy",
        band="broadband",
    )
    ms_std = group_channel_std_map(
        canonical,
        info,
        group="ms",
        condition=condition,
        measure="spectral_entropy",
        band="broadband",
    )
    shared_limit = max(float(hc_std.max()), float(ms_std.max()))
    std_difference = ms_std - hc_std
    difference_limit = max(
        abs(float(std_difference.min())),
        abs(float(std_difference.max())),
        1e-12,
    )

    figure, axes = plt.subplots(1, 3, figsize=(14, 4.5))
    plot_topomap(
        hc_std,
        info,
        title=f"HC - {condition} - SD tra soggetti",
        ax=axes[0],
        cmap="magma",
        vlim=(0.0, shared_limit),
    )
    plot_topomap(
        ms_std,
        info,
        title=f"MS - {condition} - SD tra soggetti",
        ax=axes[1],
        cmap="magma",
        vlim=(0.0, shared_limit),
    )
    plot_topomap(
        std_difference,
        info,
        title="MS − HC - differenza SD",
        ax=axes[2],
        cmap="RdBu_r",
        vlim=(-difference_limit, difference_limit),
    )
    figure.suptitle("Spectral entropy - variabilità tra soggetti")
    figure.tight_layout()
    figure.savefig(
        TOPOMAP_FIGURES/f"spectral_entropy_between_subject_sd_{condition}.png",
        dpi=200,
        bbox_inches="tight",
    )
    plt.show()
```
